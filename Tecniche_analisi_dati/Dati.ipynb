```python
# Crea una nuova colonna 'id_comune' con un ID univoco per ciascun comune
df['id_comune'] = pd.factorize(df['comune'])[0]

# Visualizza il DataFrame con la colonna ID
print(df)
```

```python
df_comuni_unici = df[['comune', 'id_comune']].drop_duplicates(subset=['comune'])
print(df_comuni_unici)
```

```python

# Ottieni gli ID univoci per ciascun comune
comuni_unici = df['comune'].drop_duplicates().reset_index(drop=True)

# Crea un dizionario che mappa ciascun comune al suo ID
comuni_id = {comune: idx for idx, comune in enumerate(comuni_unici)}

# Aggiungi una nuova colonna 'id_comune' nel DataFrame utilizzando il dizionario
df['id_comune'] = df['comune'].map(comuni_id)

print(df['id_comune'])
```

```python
# Esegui una join interna al DataFrame (auto-join) usando 'id_comune'
df_joined = df.merge(df, on='id_comune', how='left', suffixes=('_left', '_right'))

# Visualizza il DataFrame risultante
print(df_joined)
```
